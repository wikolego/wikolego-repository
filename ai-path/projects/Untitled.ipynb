```python
from matplotlib import pyplot as plt

# df["volatile acidity"].plot()

x_data = df2[df2.columns[11]]
y_data = df2[df2.columns[1]]

plt.scatter(x_data, y_data)
plt.show()
```

```python
fig, ax = plt.subplots(nrows=2, ncols=5, figsize=(18, 6))

columns = df2.columns[:]

captured_index = 10
y_data = df2[df2.columns[captured_index]]

index = 0
for r in range(2):
    for c in range(5):
        if index == captured_index:
            index += 1
        column_name = columns[index]
        x_data = df2[column_name]
        ax[r][c].set_title(f"{index} - {column_name}")
        ax[r][c].scatter(x_data, y_data)
        index += 1
        
fig.subplots_adjust(hspace=0.4, wspace=0.2)
plt.suptitle(columns[captured_index] + " compared to:")
plt.show()



# interesting relations:

# fixed acidity - density
# fixed acidity - pH
# fixed acidity - critic acid
# free sulfur dioxide - total sulfur dioxide
```
