```python
import numpy as np
import pandas as pd
```

```python
data = pd.read_csv("./data/titanic.csv")
```

```python
data.drop([460, 462], axis=0, inplace=True)
```

```python
data[data["ilość win"].notnull()]
# data = data[data["ilość win"].isnull()]
```

```python
data.drop(["ilość win", "link", "wysyłka w ciągu", "wino koszerne"], axis=1, inplace=True)
```

```python
# data["name"]
def convertName(name:str) -> str:
    valid_word = []
    for word in name.split():
        if word in ['ml', 'Ml', 'ML', 'l', 'L', 'litrów', 'NV']:
            continue
        if any(char.isdigit() for char in word):
            continue
        valid_word.append(word)

    return " ".join(valid_word)

data["name"] = data["name"].transform(convertName)
# convertName()
```

```python
data["price"] = data["price"].str.rstrip(" zł")
data["price"] = data["price"].str.replace(",", ".")
data["price"] = data["price"].str.replace("\xa0", "")
data["price"] = data["price"].astype("float64")
```

```python
# data[["rocznik", "rocznik2", "rocznik3"]].info()

data[(data["rocznik2"].notnull()) | (data["rocznik3"].notnull())][["rocznik", "rocznik2", "rocznik3"]]
```

```python
def convertVintage(row:pd.Series) -> pd.Series:
    vintages = []
    for vintage in row[["rocznik", "rocznik2", "rocznik3"]]:
        try:
            vintage = int(vintage)
        except:
            continue
        vintages.append(vintage)

    row["rocznik"] = np.mean(vintages)

    return row

data = data.transform(convertVintage)
```
