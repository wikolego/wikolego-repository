# Line/Area plot

```python
# energy

fig, ax = plt.subplots(nrows=2, ncols=3, figsize=(12, 6))

columns = energy.columns[2:]

for r in range(2):
    for c in range(3):
        column_name = columns[r*3 + c]
        ax[r][c].plot(np.arange(len(energy)), energy["All"])
        ax[r][c].fill_between(np.arange(len(energy)), np.zeros(len(energy)), energy["All"], color="lightgray")
        ax[r][c].set_ylim(0, max(energy["All"]))
        ax[r][c].set_title(column_name)
        ax[r][c].plot(np.arange(len(energy)), energy[column_name])
        ax[r][c].fill_between(np.arange(len(energy)), np.zeros(len(energy)), energy[column_name], color="red")

fig.subplots_adjust(wspace=0.4, hspace=0.35)
plt.show()
```
