# Libraries

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
# import seaborn as sns

import warnings
warnings.filterwarnings("ignore")
```

# Data

```python
data = pd.read_csv("./data/titanic.csv")
data.head()
```

# Histogram

```python
plt.figure(figsize=(10, 5))
plt.hist(data["Age"], bins=30, color='red')
plt.xticks(np.arange(0, 71, 10), [f"{age} yrs." for age in np.arange(0, 71, 10)])
plt.xlabel("Age")
plt.title("Age distribution")
plt.show()
```
