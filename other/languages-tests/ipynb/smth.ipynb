```python
import numpy as np
import matplotlib.pyplot as plt

# Data for the graph
# x = [1, 2, 3, 4, 5]  # X-axis values
# y = [1, 4, 9, 16, 25]  # Y-axis values

x = np.linspace(-10, 10, 2000)  # Generate 100 points between 0 and 10

# Create Y-axis values (e.g., y = x^2)
y = x / (np.e ** ((x ** 2) / 4))

# Create the plot
plt.plot(x, y, label="y = x^2", color='blue')

# Add title and labels
plt.title("Example Line Plot")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")

# Add a legend
plt.legend()

# Show the graph
plt.show()
```
